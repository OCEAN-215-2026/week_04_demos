## Numpy 1D array creation

````python
# creating numpy array using np.arange()
# Note that np.arange is end-point exclusive

np.arange(1, 10, 1) # start = 1, end = 10, step = 1
````

````python
# creating numpy array using np.linspace()
# Note that np.linspace is endpoint inclusive

np.linspace(1, 10, 10) # start = 1, end = 10, array length = 10
````

````python
# create an array of zero using np.zero()

np.zeros(6)
````

````python
# create an array of 7 using np.full()

np.full(3, 7)
````

````python
# You can also specify shape indirectly by referencing another array

A = np.array([2, 3, 5, 7])
np.zeros_like(A)
````

## Numpy operations

````python
# python lists, where `+` is concatenation
list_1 = [1, 2, 3, 4, 5]
list_2 = [19, 17, 15, 13, 11]

list_1 + list_2
````

````python
# you can also use `+` between a scalar and an array
array_1 + 9
````

````python
# To join two arrays, use np.concatenate()

np.concatenate([array_1, array_2])
````

````python
# more element-wise binary opeartions (*)
array_1 * array_2
````

````python
# more element-wise binary opeartions (-)
array_2 - array_1
````

````python
# more element-wise binary opeartions (**)
array_2 ** array_1
````

## Numpy indexing and slicing

````python
array = np.arange(0, 101, 10)
array
````

````python
# extracting the fourth (index = 3) element
array[3]
````

````python
# extracting the fourth to the eighth element
array[3:8]
````

````python
# extracting element from first to the eighth in step of 2
array[:8:2]
````

### Indexing with boolean array

````python
# Remainder upon division by 20 
remainder = (array % 20)
remainder
````

````python
# True if the corresponding element is divisible by 20
divisible = remainder == 0
````

````python
# extracting only the elements that are divisible by 20
array[divisible]
````

````python
# Doing the above in a single step
array[(array % 20) == 0]
````

## Missing/invalid value (`np.nan`) in numpy

````python
# np.nan can be generated by undefined numerical operations
A = np.array([0, 1, 2])
B = np.array([0, 0, 1])

X = A/B
X
````

````python
# More likely, it is used to represent missing value in data
# Example: sea surface temperature (in °C), missing value coded as -999
temp = np.array([17.2, 18.4, -999, 16.5, -999])
temp[temp < -990] = np.nan
temp
````

````python
# np.nan values generally propagates in numerical operations
temp_F = temp * 9/5 + 32.0
temp_F
````

````python
# another (unfortunate) example of np.nan propagation
np.mean(temp)
````

````python
# to filter out np.nan, use np.isnan()
temp_filtered = temp[~np.isnan(temp)]
temp_filtered
````

````python
# another way is to use the nan...() version of reduction function
np.nanmean(temp)
````

## Numpy functions call signatures

Example: `np.mean()`

> numpy.mean(a, axis=None, dtype=None, out=None, keepdims=&lt;no value>, *, where=&lt;no value>)

Full documentation: https://numpy.org/doc/stable/reference/generated/numpy.mean.html

````python
A = np.arange(5)
print(A)
````

````python
# good (no error)
# All optional arguments take their default values
# notice the dtype of the output has changed

np.mean(A)
````

````python
# bad (a required argument is missing
np.mean(axis=0)
````

````python
# good (no error)

# using position only call style
# supply a subset of default values
# a = A, axis=0, dtype=int
np.mean(A, 0, int)
````

````python
# bad: keyword argument before position ones

np.mean(axis=0, A)
````

````python
# bad: positional argument out of order
np.mean(int, A, 0)
````

````python
# OK: using keyword-based call style
# a = A, axis=0, dtype=int
np.mean(axis = 0, a = A, dtype = int)
````

````python
# good (all argument specified using keyword)
np.mean(
    a = A, axis = 0, dtype = int, 
    out = None, keepdims = False, 
    where=[True, True, True, False, False]
)
````

````python
# OK (keyword-only argument supplied using keyword; all other arguments infer by position)
# But note that this is harder to READ than the previous example
np.mean(A, 0, int, None, False, where=[True, True, True, False, False])
````

````python
# bad (try to supply keyword-only argument by position)
np.mean(A, 0, int, None, False, [True, True, True, False, False])
````

## Statistics using numpy

````python
heights = np.array([
    177, 169, 182, 173, 171, 179, 179, 171, 175, 163, 
    180, 179, 179, 164, 173, 175, 177, 176, 181, 167, 
    173, 171, 175, 177, 181, 172, 178, 165, 174, 180, 
    173, 176, 180, 180, 179, 174, 175, 173, 179, 186, 
    175, 172, 175, 164, 176, 170, 174, 175, 178, 176
]) # Data of young male adult heights (in cm)

print(heights.size)
````

````python
# mean
np.mean(heights)
````

````python
# median
np.median(heights)
````

````python
# standard deviation
np.std(heights)
````

````python
# interquartile range
np.quantile(heights, 0.75) - np.quantile(heights, 0.25)
````
